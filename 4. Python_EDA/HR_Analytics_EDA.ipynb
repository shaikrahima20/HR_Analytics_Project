# 7. Exploratory Data Analysis (EDA)

Analyzing employee demographics, attrition patterns, salary trends, and workforce distribution through visualizations.

# 7.1 Employee Attrition Distribution

Visualizing the number of employees who stayed and left the organization.

```python
plt.figure(figsize=(6,4))
sns.countplot(x='Attrition', data=df)

plt.title("Employee Attrition Distribution")
plt.xlabel("Attrition")
plt.ylabel("Employee Count")

plt.show()
```

# 7.2 Department-wise Employee Distribution

Analyzing the number of employees across different departments.

```python
plt.figure(figsize=(8,5))
sns.countplot(y='Department', data=df, order=df['Department'].value_counts().index)

plt.title("Department-wise Employee Distribution")
plt.xlabel("Employee Count")
plt.ylabel("Department")

plt.show()
```

# 7.3 Gender Distribution

Visualizing the gender distribution of employees.

```python
plt.figure(figsize=(6,4))
sns.countplot(x='Gender', data=df)

plt.title("Gender Distribution")
plt.xlabel("Gender")
plt.ylabel("Employee Count")

plt.show()
```

# 7.4 Job Role Distribution

Analyzing employee count across different job roles.

```python
plt.figure(figsize=(10,6))
sns.countplot(y='JobRole', data=df, order=df['JobRole'].value_counts().index)

plt.title("Job Role Distribution")
plt.xlabel("Employee Count")
plt.ylabel("Job Role")

plt.show()
```

# 7.5 Monthly Income Distribution

Understanding the distribution of employee monthly income.

```python
plt.figure(figsize=(8,5))
plt.hist(df['MonthlyIncome'], bins=20)

plt.title("Monthly Income Distribution")
plt.xlabel("Monthly Income")
plt.ylabel("Frequency")

plt.show()
```

# 7.6 Age Distribution

Analyzing the age distribution of employees.

```python
plt.figure(figsize=(8,5))
plt.hist(df['Age'], bins=15)

plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")

plt.show()
```

# 8. Bivariate Analysis

Analyzing the relationship between employee attrition and key HR factors.

# 8.1 Attrition by Department

Analyzing employee attrition across different departments.

```python
plt.figure(figsize=(8,5))
sns.countplot(x='Department', hue='Attrition', data=df)

plt.title("Attrition by Department")
plt.xlabel("Department")
plt.ylabel("Employee Count")
plt.xticks(rotation=15)

plt.show()
```

# 8.2 Attrition by Gender

Comparing attrition between male and female employees.

```python
plt.figure(figsize=(6,4))
sns.countplot(x='Gender', hue='Attrition', data=df)

plt.title("Attrition by Gender")
plt.xlabel("Gender")
plt.ylabel("Employee Count")

plt.show()
```

# 8.3 Attrition by Overtime

Analyzing the impact of overtime on employee attrition.

```python
plt.figure(figsize=(6,4))
sns.countplot(x='OverTime', hue='Attrition', data=df)

plt.title("Attrition by Overtime")
plt.xlabel("OverTime")
plt.ylabel("Employee Count")

plt.show()
```

# 8.4 Attrition by Job Satisfaction

Exploring the relationship between job satisfaction and employee attrition.

```python
plt.figure(figsize=(7,5))
sns.countplot(x='JobSatisfaction', hue='Attrition', data=df)

plt.title("Attrition by Job Satisfaction")
plt.xlabel("Job Satisfaction")
plt.ylabel("Employee Count")

plt.show()
```

# 8.5 Attrition by Work-Life Balance

Analyzing how work-life balance relates to employee attrition.

```python
plt.figure(figsize=(7,5))
sns.countplot(x='WorkLifeBalance', hue='Attrition', data=df)

plt.title("Attrition by Work-Life Balance")
plt.xlabel("Work-Life Balance")
plt.ylabel("Employee Count")

plt.show()
```

# 8.6 Attrition by Marital Status

Comparing attrition across different marital status groups.

```python
plt.figure(figsize=(7,5))
sns.countplot(x='MaritalStatus', hue='Attrition', data=df)

plt.title("Attrition by Marital Status")
plt.xlabel("Marital Status")
plt.ylabel("Employee Count")

plt.show()
```

# 9. Correlation Analysis

Analyzing the correlation between numerical features to identify relationships and patterns.

```python
# Select numerical columns
numerical_df = df.select_dtypes(include=['int64', 'float64'])

# Correlation matrix
correlation = numerical_df.corr()

plt.figure(figsize=(14,10))
sns.heatmap(correlation, annot=True, cmap='coolwarm', fmt=".2f")

plt.title("Correlation Heatmap")

plt.xticks(rotation=45, ha='right')
plt.yticks(rotation=0)

plt.tight_layout()

plt.show()
```

# 10. Top 10 Highest Paid Employees

Identifying the employees with the highest monthly income.

```python
top_salary = df.nlargest(10, 'MonthlyIncome')

top_salary[['EmployeeNumber', 'JobRole', 'Department', 'MonthlyIncome']]
```

# 11. Average Monthly Income by Department

Comparing the average monthly income across different departments.

```python
dept_salary = df.groupby('Department')['MonthlyIncome'].mean().sort_values(ascending=False)

dept_salary
```

Visualization

```python
plt.figure(figsize=(8,5))

dept_salary.plot(kind='bar')

plt.title("Average Monthly Income by Department")
plt.xlabel("Department")
plt.ylabel("Average Monthly Income")

plt.xticks(rotation=0)

plt.show()
```

# 12. Average Monthly Income by Job Role

Analyzing the average monthly income across different job roles.

```python
job_salary = df.groupby('JobRole')['MonthlyIncome'].mean().sort_values(ascending=False)

job_salary
```

Visualization

```python
plt.figure(figsize=(10,6))

job_salary.plot(kind='bar')

plt.title("Average Monthly Income by Job Role")
plt.xlabel("Job Role")
plt.ylabel("Average Monthly Income")

plt.xticks(rotation=45, ha='right')

plt.tight_layout()

plt.show()
```
